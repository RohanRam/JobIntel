# 02. Cleaning & Exploratory Data Analysis
Analyzes cleaned job postings, demand across roles, locations, and salary distributions.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_parquet("../data/processed/jobs_clean.parquet")
print("Clean Postings:", len(df))
df.head(3)
```

```python
# Top Role Families
df["role_family"].value_counts().plot(kind="barh", figsize=(10, 5), title="Demand by Role Family")
plt.xlabel("Posting Count")
plt.show()
```

```python
# Disclosed Salary Distribution (LPA)
sal_lpa = df[df["salary_disclosed"]]["salary_midpoint_lpa"]
plt.figure(figsize=(9, 4))
sal_lpa.plot(kind="hist", bins=30, xlim=(0, 35), title="Salary Distribution (LPA)")
plt.xlabel("LPA (Lakhs per Annum)")
plt.show()
```

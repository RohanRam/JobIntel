# 01. Raw Dataset Audit & Schema Inspection
This notebook inspects the raw Indian Job Market Dataset 2025-2026.


```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/indian_job_market.csv", low_memory=False)
print("Shape:", df.shape)
df.head(3)
```

```python
print("=== Missing Values ===")
print(df.isna().sum().sort_values(ascending=False))

print("\nExact duplicates:", df.duplicated().sum())
```

```python
sal_disclosed = (df["minimumSalary"] > 0) & (df["maximumSalary"] > 0)
print(f"Salary disclosed: {sal_disclosed.sum()} ({sal_disclosed.mean()*100:.2f}%)")
```
